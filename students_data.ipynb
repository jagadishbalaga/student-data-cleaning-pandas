```python
# import library
import pandas as pd
```

```python
# import data 
df = pd.read_csv("students_data.csv")
```

```python
df.head()
```

```python
# find the duplicates
df.duplicated()
```

```python
# remove the duplicates
df.drop_duplicates()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated()
```

```python
# find the null values
df.isnull().sum()
```

```python
df['name'].str.lower().str.title()
```

```python
df.loc[:,'name'] = df['name'].str.lower().str.title()
```

```python
df
```

```python
df['age']
```

```python
df['age'].unique()
```

```python
age_mean = df['age'].mean().round(0)
age_mean
```

```python
df['age'] = df['age'].fillna(age_mean)
```

```python
df['age'].isnull().sum()
```

```python
df['gender'].str.lower().str.title()
```

```python
df.loc[:,'gender'] = df['gender'].str.lower().str.title()
```

```python
df
```

```python
df['grade'].isnull().sum()
```

```python
df['grade'].unique()
```

```python
df['grade'] = df['grade'].replace({'10th':'10','11th':'11'})
```

```python
df['grade'].unique
```

```python
df['math_score'].isnull().sum()
```

```python
math = df['math_score'].mean()
math
```

```python
df['math_score'] = df['math_score'].fillna(math)
```

```python
df.head()
```

```python
df['english_score'].isnull().sum()
```

```python
df['english_score'].unique()
```

```python
df['english_score'].dtype
```

```python
df['english_score'] = pd.to_numeric(df['english_score'],errors='coerce')
```

```python
df['english_score'].dtype
```

```python
english = df['english_score'].mean().round(0)
```

```python
df['english_score'] = df['english_score'].fillna(english)
```

```python
df
```

```python
df['enrolled_date'].unique()
```

```python
dict_date = {
    '10-06-2022':'2022-06-10',
    '06/12/2022':'2022-06-12',
    '2022/06/11':'2022-06-11'
}
```

```python
df['enrolled_date'] = df['enrolled_date'].replace(dict_date)
```

```python
df
```

```python
df['remarks'].unique
```

```python
df['remarks'] = df['remarks'].str.lower().str.title()
```
